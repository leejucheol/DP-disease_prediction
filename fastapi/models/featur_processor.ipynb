```python
# 2️⃣ UniProt 기준 전체 단백질 ID 수집
df = df.dropna(subset=['UniProt_ID'])
all_proteins = sorted(df['UniProt_ID'].unique())
protein_to_idx = {prot: idx for idx, prot in enumerate(all_proteins)}

# 3️⃣ 엣지 구성 (protein1, protein2 → UniProt_ID 매핑)
ppi_edges = df[['protein1', 'protein2']].dropna()
mapping_dict = dict(zip(df['protein1'], df['UniProt_ID']))
ppi_edges['p1'] = ppi_edges['protein1'].map(mapping_dict)
ppi_edges['p2'] = ppi_edges['protein2'].map(mapping_dict)
ppi_edges = ppi_edges.dropna()

edge_index = torch.tensor([
    [protein_to_idx[p1], protein_to_idx[p2]]
    for p1, p2 in ppi_edges[['p1', 'p2']].values
    if p1 in protein_to_idx and p2 in protein_to_idx
], dtype=torch.long).t()
```
